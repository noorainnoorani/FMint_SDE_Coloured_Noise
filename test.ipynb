```python
import tensorflow as tf

# data_dir = '/cmlscratch/margot98/learn2correct/data_preparation/nv_data'
data_dir = 'data_preparation/nv_data'

# file1 = data_dir + '/train_expo_decay_5.tfrecord'
# file2 = data_dir + '/test_expo_decay_5.tfrecord'

# file1 = data_dir + '/train_lotka_volterra_10.tfrecord'
file1 = data_dir + '/train_expo_decay_10.tfrecord'
# file3 = data_dir + '/train_mfc_gparam_hj_forward22.tfrecord'

def parse_function(example_proto):
    # Define feature description
    feature_description = {
        'equation': tf.io.FixedLenFeature([], tf.string),
        'cond_k': tf.io.FixedLenFeature([], tf.string),
        'cond_v': tf.io.FixedLenFeature([], tf.string),
        'qoi_k': tf.io.FixedLenFeature([], tf.string),
        'qoi_v': tf.io.FixedLenFeature([], tf.string),
    }

    # Parse features
    parsed_example = tf.io.parse_single_example(example_proto, feature_description)
    equation = parsed_example['equation']
    cond_k = tf.io.parse_tensor(parsed_example['cond_k'], out_type=tf.float32)
    cond_v = tf.io.parse_tensor(parsed_example['cond_v'], out_type=tf.float32)
    qoi_k = tf.io.parse_tensor(parsed_example['qoi_k'], out_type=tf.float32)
    qoi_v = tf.io.parse_tensor(parsed_example['qoi_v'], out_type=tf.float32)

    return equation, cond_k, cond_v, qoi_k, qoi_v

# Paths to TFRecord files
files = [file1]

# Create dataset
dataset = tf.data.TFRecordDataset(files)

# Map parse_function to dataset
parsed_dataset = dataset.map(parse_function)

# Iterate over the parsed dataset
for record in parsed_dataset.take(10):
    equation, cond_k, cond_v, qoi_k, qoi_v = record
    print('=====================')
    # print(equation, cond_k[0], cond_v[0])
    print(equation, cond_k.shape, cond_v.shape, qoi_k.shape, qoi_v.shape)
```
